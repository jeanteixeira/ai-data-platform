# Simple pandas transformation

This notebook creates a small sales dataset, calculates total revenue for each row, and summarizes revenue by product.

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "product": ["notebook", "keyboard", "notebook"],
        "quantity": [2, 1, 3],
        "unit_price": [12.50, 45.00, 12.50],
    }
)
sales
```

```python
revenue_by_product = (
    sales.assign(revenue=sales["quantity"] * sales["unit_price"])
    .groupby("product", as_index=False)["revenue"]
    .sum()
    .sort_values("revenue", ascending=False)
)
revenue_by_product
```
